## **1. Tipos de Dados: Classificação e Propriedades**

## **A. Dados Qualitativos (ou Categóricos)**

Descrevem uma característica ou qualidade e não são expressos numericamente, embora possam ser codificados por números.
___
### **1. Nominal**

Categorias sem ordem ou hierarquia. Não se pode dizer que uma é "maior" ou "melhor" que a outra.

**Exemplo:** Cor dos olhos (Azul, Castanho, Verde); Estado Civil (Solteiro, Casado); País de origem

### **2. Ordinal**

Categorias que possuem uma ordem ou classificação natural, mas a diferença entre os valores não é significativa ou mensurável.

**Exemplo:** Nível de satisfação (Baixo, Médio, Alto); Nível Educacional (Ensino Médio, Graduação, Mestrado).
___

## **B. Dados Quantitativos (ou Numéricos)**

Representam quantidades e são expressos por números. Permitem operações matemáticas (soma, média, etc.).
___

### **1. Discreto**

Valores que resultam de uma contagem. Geralmente são números inteiros, finitos e contáveis.

**Exemplo:** Número de filhos (0, 1, 2, ...); Número de carros vendidos em um mês.

### **2. Contínuo**

Valores que resultam de uma medição. Podem assumir qualquer valor dentro de um intervalo (incluindo frações e decimais).

**Exemplo:** Altura (1.75m, 1.80m); Peso (75.5 kg); Temperatura (25.8°C).
___

## **2. Medidas de Tendência Central:**

### **A. Média**
**O que é:**

Soma de todos os valores dividida pelo número total de observações.

**Para que serve:**

Representa um conjunto de dados por um único número, mostrando o valor típico ou central daquele conjunto.

```python
# Lista de valores
valores = [10, 15, 30, 45]

# Criando uma Series do pandas
series = pd.Series(valores)

# Calculando a média com pandas
media = series.mean()

# Mostrando o resultado
print(media)
```

### **B. Mediana**

**O que é:**

Valor que fica exatamente no meio de um conjunto de dados quando eles estão ordenados do menor para o maior.
Se houver quantidade par de valores, é a média dos dois valores centrais.

**Para que serve:**

Representa o valor central real dos dados, sendo mais resistente a valores muito altos ou muito baixos (outliers) do que a média.
Mostra melhor o comportamento típico quando os dados estão muito espalhados ou têm extremos.

```python
# Lista de valores
valores = [10, 15, 30, 45]

# Criando uma Series do pandas
serie = pd.Series(valores)

# Calculando a mediana com pandas
mediana = serie.median()

# Mostrando o resultado
print(mediana)
```

### **C. Moda**

**O que é:**

O valor que aparece com maior frequência em um conjunto de dados.
Pode haver mais de uma moda (bimodal, multimodal) ou nenhuma (se todos os valores forem diferentes).

**Para que serve:**

Identificar o valor mais comum ou frequente em um conjunto, útil para dados categóricos ou quando se quer saber qual valor se repete mais.

```python
# Lista de valores
valores = [10, 15, 15, 30, 45, 45, 45]

# Criando uma Series do pandas
series = pd.Series(valores)

# Calculando a moda com pandas
moda = series.mode()

# Mostrando o resultado
print(moda)
```

## **3. Medidas de Dispersão:**

### **A. Amplitude**

**O que é:**

É a diferença entre o maior e o menor valor em um conjunto de dados.

$\text{Amplitude} = \text{Valor Máximo} - \text{Valor Mínimo}$

**Para que serve:**

Permite que você tenha uma ideia instantânea de quão longe os valores mais extremos estão uns dos outros.

```python
# Lista de valores
valores = [10, 15, 15, 30, 45, 45, 45, 80]

# Criando uma Series do pandas
series = pd.Series(valores)

# Calucalando os valores extremos
valor_maximo = series.max()   # 80
valor_minimo = series.min()   # 10

# Calculando a Amplitude
amplitude = valor_maximo - valor_minimo # 70

# Mostrando o resultado
print(amplitude)
```

### **B. Variância**

**O que é:**

É uma medida fundamental em estatística que quantifica o quão dispersos ou espalhados estão os dados em relação à sua média (valor central)

Em termos simples:

*   **Baixa variância** indica que a maioria dos pontos de dados está próxima da média.
*   **Alta variância** indica que os pontos de dados estão amplamente distribuídos e distantes da média

**Para que serve:**

Crucial para entender a **distribuição dos dados** e são a base para outras medidas estatísticas importantes, como o **desvio padrão** (que é a raiz quadrada da variância)



```python
# Lista de valores (Ex: Idades de um pequeno grupo de clientes)
idades = [25, 30, 35, 40, 50]

# Criando uma Series do pandas
series_idades = pd.Series(idades)

# Calculando a Média
media = series_idades.mean()

# Calculando a Variância
variancia = series_idades.var()

# Mostrando o resultado
print(f'Média: {media}')
print(f'Variância: {variancia}')
```

Neste exemplo, a Variância de $92.00$ é a média das distâncias quadráticas em torno da média de $36.00$.

Para uma melhor visualização e interpretação dos dados, utiliza-se o Desvio Padrão, pois ele está na mesma unidade dos valores originais e facilita o entendimento da dispersão dos dados.

### **C. Desvio Padrão**

**O que é:**

Assim como a variância, o desvio padrão é uma medida estatística que indica o quanto os valores de um conjunto de dados **se afastam da média**. Seu diferencial está no fato de **facilitar a visualização e a interpretação da dispersão dos dados**, pois é expresso na mesma unidade dos **valores originais**.

**Pra que serve:**

O desvio padrão é utilizado para compreender a distribuição dos dados e analisar sua precisão em relação à média, desempenhando um papel semelhante ao da variância, porém com uma **interpretação mais intuitiva.**

```python
# Lista de valores (Ex: Idades de um pequeno grupo de clientes)
idades = [25, 30, 35, 40, 50]

# Criando uma Series do pandas
series_idades = pd.Series(idades)

# Calculando a Média
media = series_idades.mean()

# Calculando o Desvio Padrão
desvio_padrao = series_idades.std()

# Mostando o resultado
print(f'Média: {media}')
print(f'Desvio Padrão: {desvio_padrao}')
```

Nesse caso o **Desvio Padrão** de $9.22$ anos nos diz que, em média, as idades dos clientes se desviam em cerca de $9.22$ anos da idade média ($36.00$), tornando a visualização da dispersão dos dados mais clara e intuitiva.

### **Observações:**

A **variância funciona como um passo intermediário** para o cálculo e a compreensão do **desvio padrão**, pois mede a dispersão dos dados de forma **menos intuitiva**. Para aprender o desvio padrão, geralmente calculamos manualmente a variância primeiro e, em seguida, extraímos sua **raiz quadrada**.

No entanto, quando o objetivo é analisar ou medir diretamente a dispersão dos dados, recomenda-se utilizar o **desvio padrão**, pois ele oferece uma **interpretação mais clara**, está na **mesma unidade dos valores originais** e, com ferramentas como o **Pandas, seu cálculo é feito automaticamente**, facilitando a análise dos dados.

## **4. Posição e Forma:**

### **A. Quartis**

**O que é:**

Dividem os dados em **4 partes iguais**.

| Medida                 | Percentil | Definição                                      |
|------------------------|-----------|------------------------------------------------|
| Mínimo                 | 0%        | O menor valor no conjunto de dados.           |
| Primeiro Quartil (Q1)  | 25%       | 25% dos dados estão abaixo deste valor.       |
| Mediana (Q2)           | 50%       | 50% dos dados estão abaixo deste valor.       |
| Terceiro Quartil (Q3)  | 75%       | 75% dos dados estão abaixo deste valor.       |
| Máximo                 | 100%      | O maior valor no conjunto de dados.           |

___

**Pra que serve:**

Busca entender como os valores estão distribuídos, serve para identificar onde os dados se concentram, como eles se espalham e se existem valores muito altos ou muito baixos, permitindo analisar o comportamento geral do conjunto de forma mais clara e precisa.

```python
# Lista de valores (Ex: Pontuações de um teste para 20 alunos)
pontuacoes = [60, 65, 70, 72, 75, 75, 78, 80, 80, 82, 85, 85, 88, 90, 92, 95, 95, 98, 100, 100]

# Criando uma Series do pandas
series_pontuacoes = pd.Series(pontuacoes)

# Os percentis 0.25, 0.50 e 0.75 correspondem a Q1, Q2 e Q3, respectivamente.
q1 = series_pontuacoes.quantile(0.25) # 25%
q2 = series_pontuacoes.quantile(0.50) # 50%
q3 = series_pontuacoes.quantile(0.75) # 75%

# Mostando o resultado
print(f'Q1: {q1}')
print(f'Q2: {q2}')
print(f'Q3: {q3}')
```

### **B. Percentis**

**O que é:**

Dividem os dados em **100 partes iguais**, usando porcentagens de **1% até 100%.**

**Pra que serve:**

**Percentis** dividem os dados em **100 partes iguais** e mostram a posição dos valores em qualquer porcentagem. Diferente dos **quartis**, que dividem os dados em apenas **4 partes principais** (25%, 50% e 75%), os percentis oferecem uma **visão mais detalhada da distribuição**. Eles ajudam a entender onde os dados se **concentram**, como se espalham e se **há valores muito altos ou baixos**, facilitando a análise do conjunto com clareza.

```python
# Lista de valores (Ex: Pontuações de um teste para 20 alunos)
pontuacoes = [60, 65, 70, 72, 75, 75, 78, 80, 80, 82, 85, 85, 88, 90, 92, 95, 95, 98, 100, 100]

# Criando uma Series do pandas
series_pontuacoes = pd.Series(pontuacoes)

# Calculando percentis específicos
p10 = series_pontuacoes.quantile(0.10)  # 10%
p30 = series_pontuacoes.quantile(0.30)  # 30%
p60 = series_pontuacoes.quantile(0.60)  # 60%
p90 = series_pontuacoes.quantile(0.90)  # 90%

# Mostando o resultado
print(f'10º Percentil: {p10}')
print(f'30º Percentil: {p30}')
print(f'60º Percentil: {p60}')
print(f'90º Percentil: {p90}')
```

### **C. Box Plot (Diagrama de Caixa)**

**O que é:**

O Box Plot é a ferramenta visual que **resume** essas **cinco medida**s de posição (Mínimo, Q1, Mediana, Q3, Máximo) e identifica graficamente os outliers (valores que **se destacam dos demais dados** por serem muito maiores ou muito menores que a maioria).

| Componente       | Representação                     | Significado                                                                                  |
|------------------|---------------------------------|---------------------------------------------------------------------------------------------|
| Linha Central    | Dentro da caixa                  | Representa a Mediana (Q2).                                                                  |
| Caixa            | Borda inferior (Q1) até superior (Q3) | Representa o Intervalo Interquartil (IQR), que contém os 50% centrais dos dados.            |
| "Fios" (Whiskers)| Linhas que se estendem da caixa | Vão até os valores mínimo e máximo, excluindo outliers.                                     |
| Pontos Isolados  | Pontos ou asteriscos fora dos fios | Representam os Outliers, ou seja, valores que estão a mais de 1.5 × IQR de distância de Q1 ou Q3. |

___

**Pra que serve:**

O gráfico de **boxplot** serve para visualizar rapidamente a **distribuição dos dados**, mostrando onde eles se concentram, sua **dispersão** e se existem **valores atípicos (outliers)**. Ele destaca a **mediana**, os **quartis** e os **limites dos dados**, facilitando a identificação de **assimetrias**, **variações** e possíveis **valores extremos** que podem influenciar a análise. Dessa forma, o boxplot ajuda a entender o **comportamento geral** do conjunto de dados de forma simples e eficiente.

```python
# Dados de exemplo
dados1 = [10, 12, 15, 18, 22, 25, 28]
dados2 = [5, 7, 10, 11, 12, 18, 35]  # Contém outlier
dados3 = [20, 22, 23, 24, 30]
dados4 = [8, 9, 10, 12, 13, 15, 30]
dados5 = [1, 3, 5, 7, 9, 11, 14, 15, 16]  # Valores menores

# Criando os boxplots para diferentes conjuntos de dados
plt.boxplot([dados1, dados2, dados3, dados4, dados5],
            labels=["Grupo 1", "Grupo 2", "Grupo 3", "Grupo 4", "Grupo 5"])

# Título e rótulos
plt.title("Vários Boxplots para Comparação")
plt.ylabel("Valores")

# Exibindo o gráfico
plt.show()
```

### **D. Skewness (Assimetria)**

**O que é:**

**Skewness (Assimetria)** é uma medida estatística que indica o **grau e a direção do desvio de uma distribuição em relação à simetria**. Ela mostra se os dados estão **mais concentrados à esquerda ou à direita da média**, permitindo identificar se a distribuição é **equilibrada** ou **distorcida pela presença de valores extremos (outliers)**.


| Tipo de Assimetria        | Valor do Coeficiente | Relação Média / Mediana | Característica |
|---------------------------|----------------------|-------------------------|----------------|
| Simétrica (Normal)        | ≈ 0                  | Média ≈ Mediana         | Distribuição equilibrada, com os dados espalhados igualmente em ambos os lados do centro. |
| Positiva (à Direita)      | > 0                  | Média > Mediana         | A cauda se estende para a direita. Indica a presença de outliers altos que puxam a média para cima. |
| Negativa (à Esquerda)     | < 0                  | Média < Mediana         | A cauda se estende para a esquerda. Indica a presença de outliers baixos que puxam a média para baixo. |

___

**Pra que serve:**

A análise da **Skewness** é fundamental para entender o comportamento da distribuição dos dados e tomar decisões corretas durante a análise exploratória e o pré-processamento.



*   **Escolha da Medida Central:**
Em distribuições assimétricas, a **Mediana** é mais representativa do valor típico do que a Média, pois a **média** é fortemente influenciada por outliers.
*   **Transformação de Dados:**
Muitos modelos de Machine Learning (como **Regressão Linear**) assumem que os resíduos seguem uma distribuição aproximadamente normal (simétrica).
Quando os dados apresentam alta assimetria, é comum aplicar **transformações logarítmicas** ou outras técnicas para reduzir a assimetria e melhorar o desempenho do modelo.



```python
from scipy.stats import skew # Função para calcular a assimetria


# --- 1. Geração dos Dados ---

# A. Distribuição Simétrica (Normal)
# Média ≈ Mediana ≈ Moda
dados_simetricos = np.random.normal(loc=50, scale=10, size=1000)

# B. Distribuição Assimétrica Positiva (Cauda à Direita)
# Média > Mediana (Ex: Renda ou preços de imóveis)
# Usamos a distribuição Gama para simular assimetria positiva
dados_positivos = np.random.gamma(shape=5, scale=1, size=1000) * 10

# C. Distribuição Assimétrica Negativa (Cauda à Esquerda)
# Média < Mediana (Ex: Notas de provas muito fáceis)
# Invertemos a distribuição gama e ajustamos o centro
dados_negativos = 100 - (np.random.gamma(shape=5, scale=1, size=1000) * 10)


# --- 2. Visualização e Análise ---

plt.figure(figsize=(15, 5))
plt.suptitle('Comparação Gráfica dos Tipos de Assimetria (Skewness)', fontsize=14, fontweight='bold')

# A. Gráfico Simétrico (Skew ≈ 0)
plt.subplot(1, 3, 1)
sns.histplot(dados_simetricos, kde=True, color='skyblue')
plt.title(f"A. Simétrica (Skew: {skew(dados_simetricos):.2f})")
# A Média, Mediana e Moda se sobrepõem no centro
plt.axvline(np.mean(dados_simetricos), color='r', linestyle='--', label='Média/Mediana')
plt.legend()


# B. Gráfico Assimetria Positiva (Skew > 0)
plt.subplot(1, 3, 2)
sns.histplot(dados_positivos, kde=True, color='lightcoral')
plt.title(f"B. Positiva (Skew: {skew(dados_positivos):.2f})")
# Média é puxada para a cauda (direita)
media_pos = np.mean(dados_positivos)
mediana_pos = np.median(dados_positivos)
plt.axvline(media_pos, color='r', linestyle='--', label=f'Média ({media_pos:.1f})')
plt.axvline(mediana_pos, color='b', linestyle=':', label=f'Mediana ({mediana_pos:.1f})')
plt.legend()


# C. Gráfico Assimetria Negativa (Skew < 0)
plt.subplot(1, 3, 3)
sns.histplot(dados_negativos, kde=True, color='lightgreen')
plt.title(f"C. Negativa (Skew: {skew(dados_negativos):.2f})")
# Média é puxada para a cauda (esquerda)
media_neg = np.mean(dados_negativos)
mediana_neg = np.median(dados_negativos)
plt.axvline(media_neg, color='r', linestyle='--', label=f'Média ({media_neg:.1f})')
plt.axvline(mediana_neg, color='b', linestyle=':', label=f'Mediana ({mediana_neg:.1f})')
plt.legend()


plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

### **E. Curtose (Kurtosis)**

**O que é:**

A **Curtose** é uma medida estatística que avalia a **“piqueza”** (**altura do pico**) e o **peso das caudas** de uma distribuição em relação à **Distribuição Normal**, que serve como padrão de referência.
Ela indica o quanto a distribuição é **mais ou menos propensa à ocorrência de valores extremos (outliers)**.

| Tipo de Curtose | Curtose em Excesso | Característica | Implicação |
|-----------------|-------------------|----------------|------------|
| Mesocúrtica | ≈ 0 | Semelhante à Distribuição Normal. | Dispersão e caudas dentro do padrão esperado. |
| Leptocúrtica | > 0 | Pico mais alto e caudas mais pesadas (gordas). | Maior probabilidade de outliers extremos (valores raros). |
| Platicúrtica | < 0 | Pico mais achatado e caudas mais leves (finas). | Menor probabilidade de outliers e maior uniformidade dos dados. |

___


**Pra que serve:**

A análise da **Curtose** é fundamental para compreender o **comportamento extremo dos dados**, auxiliando na avaliação de **riscos** e na **robustez dos modelos estatísticos e de Machine Learning**.
Curtoses elevadas indicam a necessidade de um **tratamento cuidadoso dos outliers**, pois esses valores tendem a ocorrer com maior frequência e podem **distorcer métricas, modelos e previsões**.

```python
from scipy.stats import norm, laplace, uniform
from scipy.stats import kurtosis # Função para calcular a curtose em excesso


# --- 1. Geração dos Dados ---

# A. Leptocúrtica (Curtose em Excesso > 0)
# Distribuição com caudas "pesadas" e pico alto (Ex: Distribuição de Laplace)
# Simula uma alta probabilidade de outliers
dados_lepto = laplace.rvs(loc=0, scale=1, size=10000)

# B. Mesocúrtica (Curtose em Excesso ≈ 0)
# Distribuição Normal (Gaussiana) - o padrão de referência
dados_meso = norm.rvs(loc=0, scale=1, size=10000)

# C. Platicúrtica (Curtose em Excesso < 0)
# Distribuição com caudas "leves" e pico achatado (Ex: Distribuição Uniforme)
# Simula baixa probabilidade de outliers
dados_plati = uniform.rvs(loc=-3, scale=6, size=10000)


# --- 2. Visualização e Análise ---

plt.figure(figsize=(15, 5))
plt.suptitle('Comparação Gráfica dos Tipos de Curtose (Kurtosis)', fontsize=14, fontweight='bold')

# A. Gráfico Leptocúrtico (Pico Fino e Caudas Gordas)
plt.subplot(1, 3, 1)
sns.histplot(dados_lepto, kde=True, color='purple', bins=50)
curtose_lepto = kurtosis(dados_lepto)
plt.title(f"A. Leptocúrtica (Excesso: {curtose_lepto:.2f} > 0)")
plt.text(-4, 900, 'Alto Risco de Outliers', color='purple')
plt.xlim(-5, 5) # Padroniza o eixo X para comparação
plt.ylim(0, 1000) # Padroniza o eixo Y


# B. Gráfico Mesocúrtico (Normal - Referência)
plt.subplot(1, 3, 2)
sns.histplot(dados_meso, kde=True, color='gray', bins=50)
curtose_meso = kurtosis(dados_meso)
plt.title(f"B. Mesocúrtica (Excesso: {curtose_meso:.2f} ≈ 0)")
plt.text(-4, 900, 'Referência (Curva Normal)', color='gray')
plt.xlim(-5, 5)
plt.ylim(0, 1000)


# C. Gráfico Platicúrtico (Pico Achatado e Caudas Finas)
plt.subplot(1, 3, 3)
sns.histplot(dados_plati, kde=True, color='teal', bins=50)
curtose_plati = kurtosis(dados_plati)
plt.title(f"C. Platicúrtica (Excesso: {curtose_plati:.2f} < 0)")
plt.text(-4, 900, 'Baixo Risco de Outliers', color='teal')
plt.xlim(-5, 5)
plt.ylim(0, 1000)


plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

print(f"\nINTERPRETAÇÃO DOS VALORES DE CURTOSE:")
print(f"Leptocúrtica (Excesso > 0): O pico é mais alto e o peso das caudas é maior, indicando que valores extremos (outliers) são mais prováveis do que na distribuição Normal.")
print(f"Platicúrtica (Excesso < 0): O pico é mais achatado e o peso das caudas é menor, indicando que os dados estão mais uniformemente dispersos, e valores extremos são menos prováveis.")
```

## **5. Distribuições Básicas:**


### **A. Distribuição Normal (Gaussiana)**

**O que é:**

A **Distribuição Normal** (também chamada de **Curva de Sino**) é um modelo estatístico que descreve como os dados se distribuem em muitos fenômenos naturais e sociais.
Ela é caracterizada por uma forma simétrica e em formato de sino, onde a maioria dos valores se concentra em torno da **média** e as probabilidades diminuem conforme nos afastamos dela.

**Pra que serve:**

A Distribuição Normal modela muitos fenômenos naturais, como altura, peso e notas. É a base de vários testes estatísticos que assumem dados normalmente distribuídos. Também permite calcular a probabilidade de um valor ocorrer em um intervalo, usando a área sob a curva.

```python
# 1. Definição dos Parâmetros da Distribuição Normal
media = 100       # Média (µ) - Centro da curva
desvio_padrao = 15 # Desvio Padrão (σ) - Largura da curva
tamanho_amostra = 10000

# 2. Geração de Dados Aleatórios com Distribuição Normal (usando NumPy)
# np.random.normal(loc=media, scale=desvio_padrao, size=tamanho_amostra)
dados_normais = np.random.normal(loc=media, scale=desvio_padrao, size=tamanho_amostra)

# 3. Visualização (Histograma e KDE)
plt.figure(figsize=(10, 6))
# sns.histplot cria um histograma e adiciona a Curva de Densidade (KDE)
sns.histplot(dados_normais, kde=True, bins=50, color='royalblue', alpha=0.6)

# Adicionando linhas para Média e ±1 Desvio Padrão para referência
plt.axvline(media, color='red', linestyle='--', label=f'Média (µ) = {media}')
plt.axvline(media + desvio_padrao, color='green', linestyle=':', label=f'µ + 1σ')
plt.axvline(media - desvio_padrao, color='green', linestyle=':', label=f'µ - 1σ')

plt.title('Distribuição Normal (Curva de Sino)')
plt.xlabel(f'Valores (µ={media}, σ={desvio_padrao})')
plt.ylabel('Frequência')
plt.legend()
plt.grid(axis='y', alpha=0.5)
plt.show()

print(f"Média Calculada: {np.mean(dados_normais):.2f}")
print(f"Desvio Padrão Calculado: {np.std(dados_normais):.2f}")
```

```python
# Célula de Código (Python)
# ------------------------------------------------------------------------------
# 1.7.3. Exemplo Completo: Distribuição Normal, Escore Z e Gráfico
# ------------------------------------------------------------------------------

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# --- PARÂMETROS DO CENÁRIO FICTÍCIO ---
# Cenário: Teste de QI (Quociente de Inteligência)
media_qi = 100       # Média (µ) padrão do QI
desvio_padrao_qi = 15 # Desvio Padrão (σ) padrão do QI
tamanho_amostra = 10000

# Valor de Interesse (X): QI de uma pessoa específica
qi_interesse = 130


# 1. Geração de Dados Aleatórios com Distribuição Normal (usando NumPy)
dados_qi = np.random.normal(loc=media_qi, scale=desvio_padrao_qi, size=tamanho_amostra)


# 2. CÁLCULOS ESTATÍSTICOS

# 2a. Cálculo do Escore Z
# Z = (X - µ) / σ
escore_z = (qi_interesse - media_qi) / desvio_padrao_qi

# 2b. Cálculo da Probabilidade (CDF - Área sob a curva)
# P(X ≤ 130): Probabilidade de ter QI de 130 ou menos.
prob_menor_ou_igual = stats.norm.cdf(escore_z)

# P(X > 130): Probabilidade de ser maior que 130 (a cauda superior).
prob_maior_que = 1 - prob_menor_ou_igual


# --- 3. VISUALIZAÇÃO E INTERPRETAÇÃO ---

plt.figure(figsize=(10, 6))
# sns.histplot cria um histograma e adiciona a Curva de Densidade (KDE)
sns.histplot(dados_qi, kde=True, bins=50, color='royalblue', alpha=0.6)

# Adicionando linhas para Média e ±1 Desvio Padrão para referência
plt.axvline(media_qi, color='red', linestyle='--', label=f'Média (µ) = {media_qi}')
plt.axvline(media_qi + desvio_padrao_qi, color='green', linestyle=':', label=f'µ + 1σ')
plt.axvline(media_qi - desvio_padrao_qi, color='green', linestyle=':', label=f'µ - 1σ')

# Adicionando a linha do QI de Interesse (X=130)
plt.axvline(qi_interesse, color='orange', linestyle='-', linewidth=2, label=f'QI de Interesse (X) = {qi_interesse}')


plt.title('Distribuição Normal de QI (Curva de Sino)')
plt.xlabel(f'Valores de QI (µ={media_qi}, σ={desvio_padrao_qi})')
plt.ylabel('Frequência')
plt.legend()
plt.grid(axis='y', alpha=0.5)
plt.show() # Exibe o gráfico da distribuição normal.


print("\n--- Resultados dos Dados Gerados ---")
print(f"Média Calculada (Amostra): {np.mean(dados_qi):.2f}")
print(f"Desvio Padrão Calculado (Amostra): {np.std(dados_qi):.2f}")
print("-" * 40)

print("--- Análise do QI de Interesse (130) ---")
print(f"1. Escore Z para QI {qi_interesse}: {escore_z:.2f}")
print(f"2. Pessoas com QI até {qi_interesse}: {prob_menor_ou_igual * 100:.2f}%")
print(f"3. Pessoas com QI maior que {qi_interesse}: {prob_maior_que * 100:.2f}%")
print("-" * 40)

# Conclusão prática (Seu ponto de interesse)
if abs(escore_z) < 3:
    raridade_status = "não é considerado um valor extremamente raro"
else:
    raridade_status = "é um valor raro (outlier)"

print("Conclusão Estatística:")
print(f"O QI {qi_interesse} (Escore Z de {escore_z:.2f}) {raridade_status}.")
print(f"Aproximadamente {prob_menor_ou_igual * 100:.2f}% das pessoas da população têm QI de até {qi_interesse}.")
```
